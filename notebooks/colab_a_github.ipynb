# De Colab a GitHub
Colab es tu **laboratorio**. GitHub es la **única versión oficial** del proyecto.

Colab borra los archivos cuando se desconecta. Lo que no está en GitHub (o en tu Drive) se pierde.

Orden: celda 1 → editar → celda 3 (revisar) → celda 4 (descargar) → **commit y push desde tu compu**.

## 1. Traer el código más reciente
Pega la dirección de tu repositorio. Funciona directo si el repositorio es **público**.

```python
#@title 1A. Repositorio público: clonar o actualizar
REPO_URL = "https://github.com/USUARIO/REPOSITORIO" #@param {type:"string"}
import os, subprocess
carpeta = "/content/" + REPO_URL.rstrip("/").removesuffix(".git").split("/")[-1]
if os.path.isdir(carpeta + "/.git"):
    orden = ["git", "-C", carpeta, "pull", "--ff-only"]
else:
    orden = ["git", "clone", REPO_URL, carpeta]
r = subprocess.run(orden, capture_output=True, text=True, env={**os.environ, "GIT_TERMINAL_PROMPT": "0"})
print(r.stdout + r.stderr)
if r.returncode == 0:
    os.chdir(carpeta)
    print("Listo. Estás en", carpeta)
    !git log -3 --format="%h  %an  %ar  %s"
else:
    print("No se pudo. Si el repositorio es privado usa la celda 1B.")
    print("Si dice 'Not possible to fast-forward' o menciona cambios locales, descarga tus cambios y reinicia el entorno.")
```

```python
#@title 1B. Repositorio privado: subir el ZIP
# En GitHub: botón verde "Code" > "Download ZIP". Luego ejecuta esta celda y elige ese archivo.
import io, os, zipfile
from google.colab import files
subido = files.upload()
z = zipfile.ZipFile(io.BytesIO(next(iter(subido.values()))))
z.extractall("/content")
carpeta = "/content/" + z.namelist()[0].split("/")[0]
os.chdir(carpeta)
print("Listo. Estás en", carpeta)
```
